```python
df = spark.read.format("parquet").load("abfss://bronze@stdehealthcareanalytics.dfs.core.windows.net/visits")
df.display()
```

```python
# removing NULL primary keys 

df_clean = df.dropna(subset=["visit_id","patient_id"])
df_clean.display()
```

```python
# CLEANING and Remove ALL weird characters
from pyspark.sql.functions import col, trim, lower, regexp_replace

df_clean = df_clean \
    .withColumn("visit_id", regexp_replace(col("visit_id"),"[^a-zA-Z0-9]","")) \
    .withColumn("patient_id", regexp_replace(col("patient_id"),"[^a-zA-Z0-9]","")) \
    .withColumn("provider_id", regexp_replace(col("provider_id"),"[^a-zA-Z0-9]",""))

df_clean.display()
```

```python
from pyspark.sql.functions import trim, upper, initcap, col

df_clean = df_clean \
    .withColumn("visit_type", initcap(trim(col("visit_type")))) \
    .withColumn("diagnosis", upper(trim(col("diagnosis")))) \
    .withColumn("visit_status", upper(trim(col("visit_status"))))
```

```python
df_clean = df_clean.dropDuplicates(["visit_id"])
```

```python
from pyspark.sql.functions import col
from delta.tables import DeltaTable
from pyspark.sql.utils import AnalysisException

silver_path = "abfss://silver@stdehealthcareanalytics.dfs.core.windows.net/visits/visits_silver"
bad_path = "abfss://silver@stdehealthcareanalytics.dfs.core.windows.net/visits/visits_bad"

# Split data
df_silver = df_clean.filter(col("visit_id").isNotNull() & col("patient_id").isNotNull())
df_error = df_clean.filter(col("visit_id").isNull() | col("patient_id").isNull())

# Deduplicate
df_silver = df_silver.dropDuplicates(["visit_id"])

# MERGE (incremental)
try:
    delta_table = DeltaTable.forPath(spark, silver_path)

    delta_table.alias("target") \
        .merge(df_silver.alias("source"), "target.visit_id = source.visit_id") \
        .whenMatchedUpdateAll() \
        .whenNotMatchedInsertAll() \
        .execute()

except AnalysisException:
    df_silver.write.format("delta").mode("overwrite").save(silver_path)

# Write bad records (append is OK here)
df_error.write.format("delta").mode("append").save(bad_path)
```
